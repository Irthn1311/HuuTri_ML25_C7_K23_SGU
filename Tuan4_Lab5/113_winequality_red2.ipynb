## Thống kê mô tả trên tập dữ liệu về phân loại chất lượng rượu vang đỏ

### Chuẩn bị môi trường và tải dữ liệu

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns

df = pd.read_csv('winequality-red.csv')
df = df[['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density', 'pH', 'sulphates', 'alcohol', 'quality']]
```

### Giới thiệu tổng quan về dữ liệu

```python
print("5 dòng đầu tiên của DataFrame:")
print(df.head())

# Hiển thị thông tin tóm tắt của DataFrame
print("\nThông tin tóm tắt của DataFrame:")
print(df.info())

# Kiểm tra các giá trị thiếu
print("\nTổng số giá trị thiếu cho mỗi cột:")
print(df.isnull().sum())
```

### Bảng tóm tắt thống kê mô tả

```python
print(" BẢNG TÓM TẮT THỐNG KÊ MÔ TẢ ")
# Sử dụng .T để chuyển vị bảng cho dễ đọc hơn
print(df.describe().T)
```

### Phân tích xu hướng trung tâm và độ phân tán

```python


data_mean = np.mean(df['fixed acidity'])

# Median
data_median = np.median(df['fixed acidity'])

# Mode
data_mode = stats.mode(df['fixed acidity'], keepdims=True)

# Variance
data_variance = np.var(df['fixed acidity'])

# Standard Deviation
data_sd = np.std(df['fixed acidity'])

# Range
data_range = np.max(df['fixed acidity']) - np.min(df['fixed acidity'])

# Percentiles
data_percentile25 = np.percentile(df['fixed acidity'], 25)
data_percentile50 = np.percentile(df['fixed acidity'], 50)
data_percentile75 = np.percentile(df['fixed acidity'], 75)

# Quartiles
data_quartile1 = np.quantile(df['fixed acidity'], 0.25)
data_quartile3 = np.quantile(df['fixed acidity'], 0.75)

# IQR
data_IQR = stats.iqr(df['fixed acidity'])

# In kết quả
print("Mean:", data_mean)
print("Median:", data_median)
print("Mode:", data_mode.mode[0])
print("Variance:", data_variance)
print("Standard Deviation:", data_sd)
print("Range:", data_range)
print("25th Percentile:", data_percentile25)
print("50th Percentile:", data_percentile50)
print("75th Percentile:", data_percentile75)
print("Quartile 1:", data_quartile1)
print("Quartile 3:", data_quartile3)
print("IQR:", data_IQR)
```
